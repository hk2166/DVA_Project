```python

try:
    import seaborn  # check if installed in current kernel
except ModuleNotFoundError:
    %pip install seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/cleaned_data.csv")
```

```python
df.shape
df.info()
df.describe()
```

### Observations
- Dataset contains 14 rows and Y columns
- Key variables include social media usage, sleep, and mental health indicators

```python
sns.regplot(x='daily_social_media_hours', y='stress_level', data=df)
plt.title("Social Media Usage vs Stress Level")
plt.show()
```

```python
df.groupby('high_social_media_usage')['stress_level'].mean()
```

### Insight
- High social media users (>4 hrs) show higher average stress levels
- Confirms behavioral threshold effect

```python
sns.boxplot(x='low_sleep', y='anxiety_level', data=df)
plt.title("Sleep vs Anxiety")
plt.show()
```

```python
df.groupby('low_sleep')['anxiety_level'].mean()
```

### Insight
- Anxiety levels do not differ significantly between sleep groups
- Suggests sleep alone may not strongly influence anxiety in this dataset

```python
sns.barplot(x='platform_usage', y='mental_health_score', data=df)
plt.title("Platform vs Mental Health Score")
plt.show()
```
